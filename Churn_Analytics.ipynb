```python
# Import the libraries
import pandas as pd
import numpy as np
```

```python
# Load the raw churn data downloaded from Kaggle (radheshyamkollipara/bank-customer-churn)
raw_file_path = "Raw_Data/Customer-Churn-Records.csv"
customer_data = pd.read_csv(raw_file_path)
print("Rows and columns:", customer_data.shape)
customer_data.head()
```

```python
# Data types and column information
customer_data.info()
```

```python
# Check missing values and duplicates
print("Missing values:", customer_data.isnull().sum().sum())
print("Duplicate rows:", customer_data.duplicated().sum())
print("Duplicate customer ids:", customer_data["CustomerId"].duplicated().sum())
```

```python
# Summary statistics of the number columns
customer_data.describe().round(2)
```

```python
# Check that the values are in a sensible range
print("Age range:", customer_data["Age"].min(), "to", customer_data["Age"].max())
print("Tenure range:", customer_data["Tenure"].min(), "to", customer_data["Tenure"].max())
print("Credit score range:", customer_data["CreditScore"].min(), "to", customer_data["CreditScore"].max())
print("Negative balances:", (customer_data["Balance"] < 0).sum())
print("Customers with zero balance:", (customer_data["Balance"] == 0).sum())
print("Products owned:", sorted(customer_data["NumOfProducts"].unique()))
print("Satisfaction scores:", sorted(customer_data["Satisfaction Score"].unique()))
print("Countries:", list(customer_data["Geography"].unique()))
print("Card types:", list(customer_data["Card Type"].unique()))
print("Churn flag values:", sorted(customer_data["Exited"].unique()))
print("Complaint flag values:", sorted(customer_data["Complain"].unique()))
print("Active flag values:", sorted(customer_data["IsActiveMember"].unique()))
print("Credit card flag values:", sorted(customer_data["HasCrCard"].unique()))
```

```python
# Drop columns we do not need and give the rest clear names
# Row number is just an index and surname is personal information
clean_data = customer_data.drop(columns=["RowNumber", "Surname"])
clean_data = clean_data.rename(columns={
    "CustomerId": "Customer ID",
    "CreditScore": "Credit Score",
    "Geography": "Country",
    "Tenure": "Tenure Years",
    "NumOfProducts": "Products Owned",
    "HasCrCard": "Credit Card Flag",
    "IsActiveMember": "Active Flag",
    "EstimatedSalary": "Estimated Salary",
    "Exited": "Churn Flag",
    "Complain": "Complaint Flag",
    "Point Earned": "Points Earned"})
print(list(clean_data.columns))
```

```python
# Add easy to read text labels next to the 0 and 1 columns
clean_data["Customer Status"] = clean_data["Churn Flag"].map({0: "Retained", 1: "Churned"})
clean_data["Activity Status"] = clean_data["Active Flag"].map({1: "Active", 0: "Inactive"})
clean_data["Credit Card Holder"] = clean_data["Credit Card Flag"].map({1: "Yes", 0: "No"})
clean_data["Complaint Filed"] = clean_data["Complaint Flag"].map({1: "Yes", 0: "No"})
clean_data["Card Type"] = clean_data["Card Type"].str.title()
clean_data[["Customer Status", "Activity Status", "Credit Card Holder", "Complaint Filed", "Card Type"]].head()
```

```python
# The data has no branch column, so each country is treated as one branch
clean_data["Branch"] = clean_data["Country"] + " Branch"
print(clean_data["Branch"].value_counts())
```

```python
# Age group
age_edges = [0, 29, 39, 49, 59, 100]
age_labels = ["Under 30", "30 to 39", "40 to 49", "50 to 59", "60 and above"]
clean_data["Age Group"] = pd.cut(clean_data["Age"], bins=age_edges, labels=age_labels)
clean_data["Age Group Order"] = clean_data["Age Group"].cat.codes + 1
```

```python
# Tenure group
tenure_edges = [-1, 1, 4, 7, 10]
tenure_labels = ["New (0 to 1 years)", "Growing (2 to 4 years)", "Established (5 to 7 years)", "Loyal (8 to 10 years)"]
clean_data["Tenure Group"] = pd.cut(clean_data["Tenure Years"], bins=tenure_edges, labels=tenure_labels)
clean_data["Tenure Group Order"] = clean_data["Tenure Group"].cat.codes + 1
```

```python
# Balance band
balance_edges = [-1, 0, 100000, 150000, 300000]
balance_labels = ["Zero Balance", "Up to 100K", "100K to 150K", "Above 150K"]
clean_data["Balance Band"] = pd.cut(clean_data["Balance"], bins=balance_edges, labels=balance_labels)
clean_data["Balance Band Order"] = clean_data["Balance Band"].cat.codes + 1
```

```python
# Credit score band and salary band
score_edges = [0, 500, 600, 700, 800, 900]
score_labels = ["500 or less", "501 to 600", "601 to 700", "701 to 800", "Above 800"]
clean_data["Credit Score Band"] = pd.cut(clean_data["Credit Score"], bins=score_edges, labels=score_labels)
clean_data["Credit Score Band Order"] = clean_data["Credit Score Band"].cat.codes + 1

salary_edges = [0, 50000, 100000, 150000, 200000]
salary_labels = ["Under 50K", "50K to 100K", "100K to 150K", "150K to 200K"]
clean_data["Salary Band"] = pd.cut(clean_data["Estimated Salary"], bins=salary_edges, labels=salary_labels)
clean_data["Salary Band Order"] = clean_data["Salary Band"].cat.codes + 1
```

```python
# Points band and product group
points_edges = [0, 400, 600, 800, 1000]
points_labels = ["Up to 400", "401 to 600", "601 to 800", "801 to 1000"]
clean_data["Points Band"] = pd.cut(clean_data["Points Earned"], bins=points_edges, labels=points_labels)
clean_data["Points Band Order"] = clean_data["Points Band"].cat.codes + 1

clean_data["Product Group"] = "Single Product"
clean_data.loc[clean_data["Products Owned"] > 1, "Product Group"] = "Multiple Products"
```

```python
# Customer value: High Value is the top 25 percent of balances, Low Value is zero balance
high_value_cutoff = clean_data["Balance"].quantile(0.75)
print("High value cutoff balance:", round(high_value_cutoff))

clean_data["Customer Value"] = "Medium Value"
clean_data.loc[clean_data["Balance"] >= high_value_cutoff, "Customer Value"] = "High Value"
clean_data.loc[clean_data["Balance"] == 0, "Customer Value"] = "Low Value"

value_order = ["High Value", "Medium Value", "Low Value"]
clean_data["Customer Value"] = pd.Categorical(clean_data["Customer Value"], categories=value_order, ordered=True)
clean_data["Customer Value Order"] = clean_data["Customer Value"].cat.codes + 1
print(clean_data["Customer Value"].value_counts())
```

```python
# Engagement segment from activity and satisfaction
clean_data["Engagement Segment"] = "Disengaged"
clean_data.loc[(clean_data["Active Flag"] == 0) & (clean_data["Satisfaction Score"] >= 3), "Engagement Segment"] = "Passive"
clean_data.loc[(clean_data["Active Flag"] == 1) & (clean_data["Satisfaction Score"] < 4), "Engagement Segment"] = "Engaged"
clean_data.loc[(clean_data["Active Flag"] == 1) & (clean_data["Satisfaction Score"] >= 4), "Engagement Segment"] = "Highly Engaged"

engagement_order = ["Highly Engaged", "Engaged", "Passive", "Disengaged"]
clean_data["Engagement Segment"] = pd.Categorical(clean_data["Engagement Segment"], categories=engagement_order, ordered=True)
clean_data["Engagement Order"] = clean_data["Engagement Segment"].cat.codes + 1
print(clean_data["Engagement Segment"].value_counts())
```

```python
# Risk score: points for each warning sign seen in the churn analysis
# Age 40 to 59 (+2), 3 or 4 products (+2), only 1 product (+1), inactive (+1),
# Germany branch (+1), female (+1), balance above zero (+1)
# Complaints are not used because almost every complaint comes from a customer who already left
clean_data["Risk Score"] = 0
clean_data.loc[(clean_data["Age"] >= 40) & (clean_data["Age"] <= 59), "Risk Score"] += 2
clean_data.loc[clean_data["Products Owned"] >= 3, "Risk Score"] += 2
clean_data.loc[clean_data["Products Owned"] == 1, "Risk Score"] += 1
clean_data.loc[clean_data["Active Flag"] == 0, "Risk Score"] += 1
clean_data.loc[clean_data["Country"] == "Germany", "Risk Score"] += 1
clean_data.loc[clean_data["Gender"] == "Female", "Risk Score"] += 1
clean_data.loc[clean_data["Balance"] > 0, "Risk Score"] += 1

risk_edges = [-1, 2, 4, 20]
risk_labels = ["Low", "Medium", "High"]
clean_data["Risk Level"] = pd.cut(clean_data["Risk Score"], bins=risk_edges, labels=risk_labels)
clean_data["Risk Level Order"] = clean_data["Risk Level"].cat.codes + 1

# Check that a higher risk level really means more churn
risk_check = clean_data.groupby("Risk Level", observed=True)["Churn Flag"].agg(["count", "mean"])
risk_check["mean"] = (risk_check["mean"] * 100).round(1)
print(risk_check)
```

```python
# Final check of the cleaned data
print("Rows in raw data:", len(customer_data))
print("Rows in clean data:", len(clean_data))
print("Missing values left:", clean_data.isnull().sum().sum())
print("Columns in clean data:", clean_data.shape[1])
print(list(clean_data.columns))
clean_data.head()
```

```python
# Q1: overall churn and retention rate
total_customers = len(clean_data)
churned_customers = clean_data["Churn Flag"].sum()
retained_count = total_customers - churned_customers
churn_rate = churned_customers / total_customers * 100
retention_rate = 100 - churn_rate

print("Total customers:", total_customers)
print("Churned customers:", churned_customers)
print("Retained customers:", retained_count)
print("Churn rate:", round(churn_rate, 1))
print("Retention rate:", round(retention_rate, 1))
```

```python
# Q2 and Q9: churn rate for each demographic segment
for segment_column in ["Gender", "Age Group", "Salary Band", "Credit Score Band", "Country"]:
    segment_summary = clean_data.groupby(segment_column, observed=True)["Churn Flag"].agg(["count", "sum", "mean"])
    segment_summary.columns = ["Customers", "Churned", "Churn Rate"]
    segment_summary["Churn Rate"] = (segment_summary["Churn Rate"] * 100).round(1)
    print(segment_summary)
    print()
```

```python
# Q3: does tenure have an impact on churn
for tenure_column in ["Tenure Years", "Tenure Group"]:
    tenure_summary = clean_data.groupby(tenure_column, observed=True)["Churn Flag"].agg(["count", "sum", "mean"])
    tenure_summary.columns = ["Customers", "Churned", "Churn Rate"]
    tenure_summary["Churn Rate"] = (tenure_summary["Churn Rate"] * 100).round(1)
    print(tenure_summary)
    print()
```

```python
# Q4 and Q5: product ownership, adoption and retention
for product_column in ["Products Owned", "Product Group", "Credit Card Holder", "Card Type"]:
    product_summary = clean_data.groupby(product_column, observed=True)["Churn Flag"].agg(["count", "sum", "mean"])
    product_summary.columns = ["Customers", "Churned", "Churn Rate"]
    product_summary["Churn Rate"] = (product_summary["Churn Rate"] * 100).round(1)
    product_summary["Retention Rate"] = (100 - product_summary["Churn Rate"]).round(1)
    print(product_summary)
    print()

# Share of customers for each number of products (adoption)
print((clean_data["Products Owned"].value_counts(normalize=True).sort_index() * 100).round(1))
```

```python
# Q6: branch wise churn and retention
branch_summary = clean_data.groupby("Branch").agg(Customers=("Customer ID", "count"),
                                                   Churned=("Churn Flag", "sum"),
                                                   Churn_Rate=("Churn Flag", "mean"),
                                                   Average_Balance=("Balance", "mean"))
branch_summary["Churn_Rate"] = (branch_summary["Churn_Rate"] * 100).round(1)
branch_summary["Retention_Rate"] = (100 - branch_summary["Churn_Rate"]).round(1)
branch_summary["Average_Balance"] = branch_summary["Average_Balance"].round(0)
print(branch_summary.sort_values("Churn_Rate", ascending=False))
```

```python
# Q7: customer activity of retained and churned customers
for activity_column in ["Activity Status", "Engagement Segment", "Satisfaction Score", "Complaint Filed"]:
    activity_summary = clean_data.groupby(activity_column, observed=True)["Churn Flag"].agg(["count", "sum", "mean"])
    activity_summary.columns = ["Customers", "Churned", "Churn Rate"]
    activity_summary["Churn Rate"] = (activity_summary["Churn Rate"] * 100).round(1)
    print(activity_summary)
    print()

print("Average activity measures by status")
print(clean_data.groupby("Customer Status")[["Active Flag", "Satisfaction Score", "Points Earned", "Products Owned"]].mean().round(2))
```

```python
# Q8: account balance, card points (spending) and churn
for band_column in ["Balance Band", "Points Band", "Customer Value"]:
    band_summary = clean_data.groupby(band_column, observed=True)["Churn Flag"].agg(["count", "sum", "mean"])
    band_summary.columns = ["Customers", "Churned", "Churn Rate"]
    band_summary["Churn Rate"] = (band_summary["Churn Rate"] * 100).round(1)
    print(band_summary)
    print()

print("Average balance by status")
print(clean_data.groupby("Customer Status")["Balance"].mean().round(0))
print()
print("Total balance lost with churned customers:", round(clean_data[clean_data["Churn Flag"] == 1]["Balance"].sum()))
```

```python
# Q10: high value customers who show signs of disengagement
high_value_customers = clean_data[clean_data["Customer Value"] == "High Value"]
print("High value customers:", len(high_value_customers))
print("Churn rate of high value customers:", round(high_value_customers["Churn Flag"].mean() * 100, 1))

high_value_retained = high_value_customers[high_value_customers["Customer Status"] == "Retained"]
disengaged_high_value = high_value_retained[high_value_retained["Activity Status"] == "Inactive"]
print("High value customers still retained:", len(high_value_retained))
print("Of these, inactive members:", len(disengaged_high_value))
print("Balance held by inactive high value customers:", round(disengaged_high_value["Balance"].sum()))
```

```python
# Q11: common patterns among churned customers
status_averages = clean_data.groupby("Customer Status")[["Age", "Balance", "Products Owned", "Active Flag",
                                                          "Credit Score", "Estimated Salary", "Satisfaction Score",
                                                          "Points Earned", "Tenure Years"]].mean().round(2)
print(status_averages)

print()
print("Share of customers in each branch (percent)")
print((pd.crosstab(clean_data["Branch"], clean_data["Customer Status"], normalize="columns") * 100).round(1))

print()
print("Share of customers by gender (percent)")
print((pd.crosstab(clean_data["Gender"], clean_data["Customer Status"], normalize="columns") * 100).round(1))
```

```python
# Q12: which segments should get retention campaigns first
# Branch and age group together, ranked by the number of customers lost
campaign_segments = clean_data.groupby(["Branch", "Age Group"], observed=True)["Churn Flag"].agg(["count", "sum", "mean"])
campaign_segments.columns = ["Customers", "Churned", "Churn Rate"]
campaign_segments["Churn Rate"] = (campaign_segments["Churn Rate"] * 100).round(1)
campaign_segments = campaign_segments.sort_values("Churned", ascending=False)
print(campaign_segments.head(8))

# Customers still with the bank who need attention, by risk level
retained_customers = clean_data[clean_data["Customer Status"] == "Retained"]
print()
print(retained_customers["Risk Level"].value_counts())
```

```python
# Q13: which factors separate churners from stayers the most
# For each factor we compare the highest and lowest churn rate between its groups
factor_columns = ["Age Group", "Products Owned", "Activity Status", "Country", "Gender", "Balance Band",
                  "Tenure Group", "Card Type", "Satisfaction Score", "Credit Score Band", "Salary Band",
                  "Points Band", "Credit Card Holder", "Complaint Filed"]

indicator_rows = []
for factor_column in factor_columns:
    factor_rates = clean_data.groupby(factor_column, observed=True)["Churn Flag"].mean() * 100
    highest_group = factor_rates.idxmax()
    lowest_group = factor_rates.idxmin()
    indicator_rows.append([factor_column, str(highest_group), round(factor_rates.max(), 1),
                           str(lowest_group), round(factor_rates.min(), 1),
                           round(factor_rates.max() - factor_rates.min(), 1)])

churn_indicators = pd.DataFrame(indicator_rows, columns=["Factor", "Highest Churn Group", "Highest Churn Rate",
                                                         "Lowest Churn Group", "Lowest Churn Rate", "Gap"])
churn_indicators = churn_indicators.sort_values("Gap", ascending=False)
print(churn_indicators)

# Complaints are filed almost only by customers who already left, so they confirm the exit
# but cannot be used to predict it early. The next strongest factors are products and age.
```

```python
# Q13 (part 2): correlation of the number columns with churn
number_columns = ["Credit Score", "Age", "Tenure Years", "Balance", "Products Owned", "Active Flag",
                  "Estimated Salary", "Complaint Flag", "Satisfaction Score", "Points Earned", "Churn Flag"]
churn_correlation = clean_data[number_columns].corr()["Churn Flag"].drop("Churn Flag")
print(churn_correlation.sort_values(ascending=False).round(3))
```

```python
# High value customers who are still with the bank but show many warning signs
at_risk_customers = clean_data[(clean_data["Customer Status"] == "Retained") &
                               (clean_data["Customer Value"] == "High Value") &
                               (clean_data["Risk Level"] == "High")]
at_risk_customers = at_risk_customers.sort_values(["Risk Score", "Balance"], ascending=False)

print("High value customers at risk:", len(at_risk_customers))
print("Balance held by them:", round(at_risk_customers["Balance"].sum()))
print(at_risk_customers[["Customer ID", "Branch", "Age", "Balance", "Products Owned", "Activity Status", "Risk Score", "Risk Level"]].head(10))
```
